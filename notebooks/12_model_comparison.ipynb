### Comprehensive Comparative Study
Compares all 14 evaluated algorithms across ROC-AUC, Balanced Accuracy, Sensitivity, Specificity, F1, and Latency.


```python
import pandas as pd

comp_df = pd.read_csv("./models/model_comparison.csv")
print(comp_df.to_string(index=False))

# Plot comparative ROC-AUC
import matplotlib.pyplot as plt
plt.figure(figsize=(10, 6))
comp_df.sort_values(by='Test_ROC_AUC', ascending=True).plot(
    kind='barh', x='Algorithm', y='Test_ROC_AUC', color='teal', legend=False
)
plt.title('Algorithm Comparison: Test ROC-AUC')
plt.xlabel('Area Under ROC Curve')
plt.xlim(0.75, 0.90)
plt.tight_layout()
plt.show()
```
