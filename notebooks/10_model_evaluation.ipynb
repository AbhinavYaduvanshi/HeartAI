### Evaluation Suite
Calculates Sensitivity (Recall), Specificity, Balanced Accuracy, Brier Score, and calibration curves across models.


```python
from sklearn.metrics import roc_curve, precision_recall_curve, brier_score_loss, confusion_matrix
from sklearn.calibration import calibration_curve
import matplotlib.pyplot as plt

# Generate evaluation curves
fig, axes = plt.subplots(1, 2, figsize=(14, 6))

# ROC Curves
fpr, tpr, _ = roc_curve(y_test, y_rf_prob)
axes[0].plot(fpr, tpr, label='Random Forest')
axes[0].plot([0, 1], [0, 1], 'k--', label='Chance')
axes[0].set_title('ROC Curve Comparison')
axes[0].set_xlabel('False Positive Rate (1 - Specificity)')
axes[0].set_ylabel('True Positive Rate (Sensitivity)')
axes[0].legend()

# Calibration Curves
prob_true, prob_pred = calibration_curve(y_test, y_rf_prob, n_bins=10)
axes[1].plot(prob_pred, prob_true, marker='o', label='Random Forest')
axes[1].plot([0, 1], [0, 1], 'k--', label='Perfect Calibration')
axes[1].set_title('Reliability Calibration Curve')
axes[1].set_xlabel('Mean Predicted Probability')
axes[1].set_ylabel('Empirical Fraction of Positives')
axes[1].legend()

plt.tight_layout()
plt.show()
```
