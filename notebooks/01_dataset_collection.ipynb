### Objective
Download the official **CDC Behavioral Risk Factor Surveillance System (BRFSS) 2023** survey dataset (433,323 respondents, 350 variables), verify file integrity, and inspect codebook target variable `_MICHD`.

> **Medical Disclaimer:** Educational machine learning research prototype. Not a clinically validated diagnostic tool.


```python
# Step 1: Environment Setup & Optional Google Drive Mount
import os
import urllib.request
import zipfile

try:
    from google.colab import drive
    drive.mount('/content/drive')
    DATA_DIR = '/content/drive/MyDrive/HeartAI/data'
except ImportError:
    DATA_DIR = './data'

os.makedirs(f"{DATA_DIR}/raw", exist_ok=True)
os.makedirs(f"{DATA_DIR}/codebook", exist_ok=True)
print(f"Working Data Directory: {DATA_DIR}")
```

```python
# Step 2: Download Official CDC BRFSS 2023 Files
DATASET_URL = "https://www.cdc.gov/brfss/annual_data/2023/files/LLCP2023XPT.zip"
CODEBOOK_URL = "https://www.cdc.gov/brfss/annual_data/2023/zip/codebook23_llcp-v2-508.zip"

print("Downloading CDC BRFSS 2023 Dataset (~93.2 MB)...")
!curl -L -o "{DATA_DIR}/raw/LLCP2023XPT.zip" "{DATASET_URL}"

print("Downloading CDC BRFSS 2023 Codebook (~3.1 MB)...")
!curl -L -o "{DATA_DIR}/codebook/codebook23_llcp.zip" "{CODEBOOK_URL}"

print("Extracting files...")
with zipfile.ZipFile(f"{DATA_DIR}/raw/LLCP2023XPT.zip", 'r') as zip_ref:
    zip_ref.extractall(f"{DATA_DIR}/raw")

with zipfile.ZipFile(f"{DATA_DIR}/codebook/codebook23_llcp.zip", 'r') as zip_ref:
    zip_ref.extractall(f"{DATA_DIR}/codebook")

print("Files successfully downloaded and verified!")
!ls -lh "{DATA_DIR}/raw"
!ls -lh "{DATA_DIR}/codebook"
```
