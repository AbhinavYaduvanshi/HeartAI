In-depth clinical exploratory analysis across demographic, hemodynamic, and electrocardiographic features.

### Objectives:
1. Identify high-leverage risk drivers for coronary artery disease.
2. Assess clinical distributions and outliers.
3. Compute bivariate correlations and risk odds ratios.


```python
# Step 1: Environment Setup & Drive Ingestion
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

try:
    from google.colab import drive
    drive.mount('/content/drive')
    DATA_PATH = '/content/drive/MyDrive/HeartAI/data/heart_train.csv'
except ImportError:
    DATA_PATH = './drive_mock/data/heart_train.csv'

if os.path.exists(DATA_PATH):
    df = pd.read_csv(DATA_PATH)
else:
    # Online fallback
    URL = "https://archive.ics.uci.edu/ml/machine-learning-databases/heart-disease/processed.cleveland.data"
    COLUMNS = ['age', 'sex', 'cp', 'trestbps', 'chol', 'fbs', 'restecg', 'thalach', 'exang', 'oldpeak', 'slope', 'ca', 'thal', 'target']
    df = pd.read_csv(URL, names=COLUMNS, na_values="?").fillna(0)
    df['target'] = (df['target'] > 0).astype(int)

print(f"Loaded clinical dataset with {df.shape[0]} patients and {df.shape[1]} features.")
```

```python
# Step 2: Target Prevalence & Demographics
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Target distribution
sns.countplot(data=df, x='target', palette=['#10b981', '#ef4444'], ax=axes[0])
axes[0].set_title('Cardiovascular Disease Prevalence (0: No CAD, 1: CAD Present)')
axes[0].set_xticklabels(['Healthy', 'Heart Disease'])

# Age distribution by cardiac status
sns.kdeplot(data=df, x='age', hue='target', fill=True, common_norm=False, palette=['#10b981', '#ef4444'], ax=axes[1])
axes[1].set_title('Age Distribution vs. Coronary Disease')
plt.tight_layout()
plt.show()
```

```python
# Step 3: Hemodynamic Risk Factors (BP, Cholesterol, Heart Rate)
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.boxplot(data=df, x='target', y='trestbps', palette=['#10b981', '#ef4444'], ax=axes[0])
axes[0].set_title('Resting Blood Pressure (mmHg)')
axes[0].set_xticklabels(['Healthy', 'Heart Disease'])

sns.boxplot(data=df, x='target', y='chol', palette=['#10b981', '#ef4444'], ax=axes[1])
axes[1].set_title('Serum Cholesterol (mg/dl)')
axes[1].set_xticklabels(['Healthy', 'Heart Disease'])

sns.boxplot(data=df, x='target', y='thalach', palette=['#10b981', '#ef4444'], ax=axes[2])
axes[2].set_title('Maximum Achieved Heart Rate (bpm)')
axes[2].set_xticklabels(['Healthy', 'Heart Disease'])

plt.tight_layout()
plt.show()
```

```python
# Step 4: Correlation Matrix Heatmap
plt.figure(figsize=(12, 10))
corr = df.corr()
mask = np.triu(np.ones_like(corr, dtype=bool))
sns.heatmap(corr, mask=mask, annot=True, fmt='.2f', cmap='coolwarm', vmin=-1, vmax=1, linewidths=0.5)
plt.title('Clinical Feature Correlation Matrix (HeartAI)')
plt.show()
```
