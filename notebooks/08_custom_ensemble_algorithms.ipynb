### Custom Algorithms
1. **Enriched Random Forest (ERF):** Weighted feature sampling using training-only mutual information relevance scores.
2. **Tree-Weighted Random Forest (TWRF):** Out-of-bag validation accuracy weighting across individual decision trees.


```python
from app.ml.custom_ensembles import EnrichedRandomForestClassifier, TreeWeightedRandomForestClassifier
from sklearn.metrics import roc_auc_score

erf = EnrichedRandomForestClassifier(n_estimators=60, max_depth=8, max_features=0.6, random_state=42)
erf.fit(X_train_proc, y_train)
print(f"Enriched Random Forest Test ROC-AUC:     {roc_auc_score(y_test, erf.predict_proba(X_test_proc)[:, 1]):.4f}")

twrf = TreeWeightedRandomForestClassifier(n_estimators=60, max_depth=8, random_state=42)
twrf.fit(X_train_proc, y_train)
print(f"Tree-Weighted Random Forest Test ROC-AUC: {roc_auc_score(y_test, twrf.predict_proba(X_test_proc)[:, 1]):.4f}")
```
