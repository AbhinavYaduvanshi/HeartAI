### Objective
Construct an enterprise-grade `ColumnTransformer` with `SimpleImputer` and `StandardScaler`. Fit transformations strictly on training folds to prevent data leakage.


```python
# Step 1: Preprocessing Pipeline Construction
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import pandas as pd
import numpy as np

# Load data
df = pd.read_csv("./data/processed/heart_disease_brfss2023.csv")

FEATURES = [c for c in df.columns if c not in ['HeartDisease', 'SurveyWeight']]
X = df[FEATURES]
y = df['HeartDisease'].values

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)
print(f"X_train shape: {X_train.shape}, X_test shape: {X_test.shape}")
```

```python
# Step 2: Fit Pipeline on Training Data Only
NUM_COLS = ["BMI", "PhysicalHealthDays", "MentalHealthDays"]
CAT_COLS = [c for c in FEATURES if c not in NUM_COLS]

preprocessor = ColumnTransformer([
    ('num', Pipeline([
        ('imputer', SimpleImputer(strategy='median')),
        ('scaler', StandardScaler())
    ]), NUM_COLS),
    ('cat', SimpleImputer(strategy='most_frequent'), CAT_COLS)
])

X_train_proc = preprocessor.fit_transform(X_train)
X_test_proc = preprocessor.transform(X_test)
print(f"Processed training array: {X_train_proc.shape}, NaNs: {np.isnan(X_train_proc).sum()}")
```
