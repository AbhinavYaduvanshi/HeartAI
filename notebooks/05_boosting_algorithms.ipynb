### Algorithms Evaluated
1. **AdaBoost**
2. **Gradient Boosting**
3. **XGBoost** (`XGBClassifier`)
4. **LightGBM** (`LGBMClassifier`)
5. **CatBoost** (`CatBoostClassifier`)


```python
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier
from sklearn.metrics import roc_auc_score

# XGBoost
xgb = XGBClassifier(n_estimators=100, learning_rate=0.08, max_depth=4, scale_pos_weight=10.0, eval_metric='logloss', random_state=42)
xgb.fit(X_train_proc, y_train)
print(f"XGBoost Test ROC-AUC:  {roc_auc_score(y_test, xgb.predict_proba(X_test_proc)[:, 1]):.4f}")

# LightGBM
lgb = LGBMClassifier(n_estimators=100, learning_rate=0.08, num_leaves=31, class_weight='balanced', random_state=42, verbose=-1)
lgb.fit(X_train_proc, y_train)
print(f"LightGBM Test ROC-AUC: {roc_auc_score(y_test, lgb.predict_proba(X_test_proc)[:, 1]):.4f}")

# CatBoost
cb = CatBoostClassifier(iterations=100, learning_rate=0.08, depth=5, auto_class_weights='Balanced', verbose=0, random_state=42)
cb.fit(X_train_proc, y_train)
print(f"CatBoost Test ROC-AUC: {roc_auc_score(y_test, cb.predict_proba(X_test_proc)[:, 1]):.4f}")
```
