### Objective
Perform comprehensive exploratory data analysis across the 397,870 usable CDC BRFSS 2023 respondents. Inspect demographic risk disparities, body mass index, blood pressure history, and target class imbalance.


```python
# Step 1: Ingest Cleaned BRFSS Dataset
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", palette="muted")

# Load processed dataset
df = pd.read_csv("https://github.com/AbhinavYaduvanshi/HeartAI/raw/main/data/processed/heart_disease_brfss2023.csv.gz"
                 if not os.path.exists("./data/processed/heart_disease_brfss2023.csv")
                 else "./data/processed/heart_disease_brfss2023.csv")

print(f"Loaded dataset: {df.shape[0]:,} rows, {df.shape[1]} columns")
print(f"Heart Disease Positive Cases: {df['HeartDisease'].sum():,} ({df['HeartDisease'].mean()*100:.2f}%)")
df.head()
```

```python
# Step 2: Distribution of Key Clinical Risk Factors
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Age Category distribution by Heart Disease status
sns.countplot(data=df, x='AgeCategory', hue='HeartDisease', ax=axes[0, 0])
axes[0, 0].set_title('Heart Disease Prevalence Across Age Categories (1: 18-24 ... 13: 80+)')

# 2. General Health rating
sns.countplot(data=df, x='GeneralHealth', hue='HeartDisease', ax=axes[0, 1])
axes[0, 1].set_title('General Health Rating (1: Excellent ... 5: Poor)')

# 3. High Blood Pressure vs Heart Disease
sns.barplot(data=df, x='HighBP', y='HeartDisease', ax=axes[1, 0])
axes[1, 0].set_title('Heart Disease Rate by High Blood Pressure Diagnosis')

# 4. BMI Distribution KDE
sns.kdeplot(data=df, x='BMI', hue='HeartDisease', common_norm=False, ax=axes[1, 1])
axes[1, 1].set_title('Body Mass Index (BMI) Distribution')

plt.tight_layout()
plt.show()
```
