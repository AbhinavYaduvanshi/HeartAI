### Objective
Packages the best performing validated model with its preprocessing pipeline and cryptographic SHA-256 checksum manifest into Google Drive for automatic Antigravity IDE synchronization.


```python
import os
import shutil
import hashlib
import json

# Mount Google Drive
try:
    from google.colab import drive
    drive.mount('/content/drive')
    EXPORT_DIR = "/content/drive/MyDrive/HeartAI/models"
except ImportError:
    EXPORT_DIR = "./drive_mock/models"

os.makedirs(EXPORT_DIR, exist_ok=True)
print(f"Export target: {EXPORT_DIR}")

# Copy best model and metadata
shutil.copy("models/best_model.joblib", f"{EXPORT_DIR}/best_model.joblib")
shutil.copy("models/preprocessing_pipeline.joblib", f"{EXPORT_DIR}/preprocessing_pipeline.joblib")
shutil.copy("models/model_metadata.json", f"{EXPORT_DIR}/model_metadata.json")

print("Model artifacts successfully transferred to Google Drive!")
```
